### Nonnegative matrix factorization
### The alternating least squares algorithm (ALS)
- Initialize $W^{(1)}$ randomly or using prior data.
- Find $H^{(1)} = \text{argmin}\left\{\|X - W^{(1)}H\|_{F}^{2}: H\in\mathbb{R}^{r\times n} \right\}$
- Until convergence do:
    - $W^{(k)} = \text{argmin}\left\{\|X - WH^{(k-1)}\|_{F}^{2}: W\in\mathbb{R}^{m\times r} \right\}$
    - $H^{(k)} = \text{argmin}\left\{\|X - W^{(k)}H\|_{F}^{2}: H\in\mathbb{R}^{r\times n} \right\}$
- Multiplicative update rule:
    - $H_{ij}^k=H_{ij}^{k-1}\frac{((W^{k-1})^TX)_{ij}}{((W^{k-1})^TW^{k-1}H^{k-1})_{ij}}$
    - $W_{ij}^k=W_{ij}^{k-1}\frac{(X(H^{k})^T)_{ij}}{(W^{k-1}H^{k}(H^{k})^T)_{ij}}$
    
 **Remarks**
 - Each sub-problem is easy (**i.e.** convex). It is essentially least squares.
 - But alternating two convex problems is no longer a convex problem.

```python
import numpy as np
```

```python
W1 = np.array([[0.162, 0.449, 0.696],[0.784, 0.0, 0.0],[0.632, 0.599, 0.0],
              [0.413, 0.0, 0.648],[0.62, 0.0, 0.0],[0.336, 0.766, 0.0]])
H1 = np.array([[0.268, 0.557, 0.896, 0.538, 0.854],
              [0.94, 0.595, 0.0, 0.0, 0.0],[0.226, 0.0, 0.0, 0.917, 0.0]])

X = np.array([[0.735, 0.179, 0.079, 0.698, 0.307],[0.093, 0.655, 0.655, 0.436, 0.604]
              ,[0.686, 0.782, 0.547, 0.427, 0.471],[0.138, 0.223, 0.573, 0.846, 0.164]
              ,[0.427, 0.018, 0.488, 0.193, 0.819],[0.78, 0.69, 0.354, 0.221, 0.184]])
print(X)
```

```python
W = np.random.rand(6, 2) # Initialize W and H as nonnegative matrices. Otherwise, the algorithm may not converge
H = np.random.rand(2, 5)
for iteration in range(20):
    H = H*(np.matmul(W.T, X)/np.matmul(W.T, np.matmul(W, H)))
    W = W*(np.matmul(X, H.T)/np.matmul(W, np.matmul(H, H.T)))
    #if iteration % 10 == 9:
    print(np.linalg.norm(X-np.matmul(W,H)))
```

```python
X = np.matmul(W1, H1) + 0.1*np.random.randn(6,5)
X = np.clip(X, 0, None)
```

```python
W = np.random.rand(6, 2) # Initialize W and H as nonnegative matrices. Otherwise, the algorithm may not converge
H = np.random.rand(2, 5)
for iteration in range(50):
    H = H*(np.matmul(W.T, X)/np.matmul(W.T, np.matmul(W, H)))
    W = W*(np.matmul(X, H.T)/np.matmul(W, np.matmul(H, H.T)))
    #if iteration % 10 == 9:
    print(np.linalg.norm(X-np.matmul(W,H)))
```
